## 3. Data Cleaning

The dataset was checked for missing values, duplicate records,
incorrect data types, and inconsistent categorical values.

The missing-value check was performed using Pandas, and duplicate
records were identified using the duplicated() function.

Since unnecessary removal of valid observations can affect the
analysis, records were not deleted without a clear reason.

The categorical variables were also checked for consistency, and
numerical variables were examined using descriptive statistics.

```python
df_clean = df.copy()

print("Cleaned dataset created successfully!")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])
```

```python
df_clean.to_csv("student-mat-cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## 4. Preliminary Exploratory Analysis

Descriptive statistics were performed to understand the central
tendency, variation, and distribution of the numerical variables
in the dataset.

```python
df_clean.describe()
```

```python
print("Average Final Grade:", round(df_clean["G3"].mean(), 2))
print("Minimum Final Grade:", df_clean["G3"].min())
print("Maximum Final Grade:", df_clean["G3"].max())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df_clean["G3"], bins=11, kde=True)

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

```python
studytime_analysis = df_clean.groupby("studytime")["G3"].mean().round(2)

print(studytime_analysis)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(x=studytime_analysis.index, y=studytime_analysis.values)

plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time Category")
plt.ylabel("Average Final Grade")

plt.show()
```

```python
absence_correlation = df_clean["absences"].corr(df_clean["G3"])

print("Correlation between Absences and Final Grade:",
      round(absence_correlation, 2))
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df_clean, x="absences", y="G3")

plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
g2_correlation = df_clean["G2"].corr(df_clean["G3"])

print("Correlation between G2 and G3:",
      round(g2_correlation, 2))
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df_clean, x="G2", y="G3")

plt.title("Previous Grade (G2) vs Final Grade (G3)")
plt.xlabel("Previous Grade (G2)")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
internet_analysis = df_clean.groupby("internet")["G3"].agg(
    ["count", "mean"]
).round(2)

print(internet_analysis)
```

```python
plt.figure(figsize=(7, 5))

sns.boxplot(data=df_clean, x="internet", y="G3")

plt.title("Internet Access vs Final Grade")
plt.xlabel("Internet Access")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
plt.figure(figsize=(12, 8))

numeric_df = df_clean.select_dtypes(include=np.number)
correlation = numeric_df.corr()

sns.heatmap(correlation, annot=True, fmt=".2f", cmap="coolwarm")

plt.title("Correlation Heatmap of Student Performance")

plt.show()
```

## Preliminary Findings

The preliminary analysis was used to identify potential relationships
between student characteristics and final academic performance.

The analysis examined final grade distribution, study time, absences,
previous academic performance, and internet access.

The average final grade provides an overall view of student performance.
The study-time analysis helps investigate whether students with
different study-time categories show differences in average final
grades.

The absence analysis provides an initial indication of the relationship
between attendance-related behavior and academic performance. The
relationship between previous grade (G2) and final grade (G3) was also
examined using correlation and scatter plots.

Internet access was analyzed by comparing final-grade distributions
between students with and without internet access.

These are preliminary observations. Further statistical analysis would
be required before treating these relationships as confirmed findings.
